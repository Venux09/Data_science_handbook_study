## Upgrading the Matplotlib visualizatoin to Searborn 

```python
 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

sns.set()
```

```python
data = np.random.multivariate_normal([0, 0], [[5, 2], [2, 2]], size=2000)
data = pd.DataFrame(data, columns=['x', 'y'])

for col in  'xy':
    plt.hist(data[col] , density = True , alpha = 0.5)
```

```python
#now plotting that with the sns for smoother version 
sns.kdeplot(data = data , shade = True
           )
```

```python
#lets try other methods with this 
```

```python
#plotting the x and y relationship between the x and y 
sns.kdeplot(data = data , x = 'x' , y = 'y')
```

```python
#let work with some well known data also know as iris dataset 
iris = sns.load_dataset('iris')
sns.pairplot(iris,hue = 'species' , height = 2.5)
```
