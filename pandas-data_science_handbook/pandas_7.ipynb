# Heirarchical indexing - Part-2

```python
import pandas as pd 
import numpy as np 
```

```python
#now we are going to do the MultiIndex for the columns 
```

```python
#creating heirarchical index for the index and columns 
index = pd.MultiIndex.from_product([[2013,2014],[1,2]],
                                  names= ['year','visit']
                                )
columns = pd.MultiIndex.from_product([['Bob','Suzy','Michael'],['HR','Temp']],
                                    names = ['Name','Type'])
```

```python
#making a fake data 
data = np.round(np.random.randn(4,6),1)#here we are making an array of 4,6 size and rounding that to 1 digit
data[:,::2] *= 10
data += 37
```

```python
#creating the data frame 
health_data = pd.DataFrame(data,index = index , columns = columns)
```

```python
health_data 
```

```python
#creating an multi index
index2 = pd.MultiIndex.from_product([['a','b','c'],[1,2]])
data2 = pd.Series(np.random.rand(6),index = index2 )
data2.index.names = ['char','int']
```

```python
data2
```

```python
try: 
    data2['a':'b']
except KeyError as e:
    print(type(e))
    print(e)
```
