```python
#function for making concated data frames 
def make_df(cols,ind):
    "Makes a quick Data frames"
    data = {c:[str(c)+str(i) for i in ind] for c in cols}
    return pd.DataFrame(data,ind)
```

```python
make_df('ABC',range(3))
```

```python
#making a class that will allow use to use multiple data frames side by side 
class display(object):
    """Display HTML representation of multiple objects"""
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}{1}
    """
    def __init__(self,*args):#storing all the arguments means variable etc 
        self.args = args

    def _repr_html_(self):
        return '\n'.join(self.template.format(a,eval(a)._repr_html_()) for a in self.args)

    def _repr_(self):
        return '\n\n'.join(a+'\n'+repr(eval(a)) for a in self.args)

    
```

```python
#duplicate indices 
#at the point pandas indices preserve even if we will make indices equal they will do the same task as it was supposed to be when it was
#with normal indexing it was doing before 
```

```python
#for example 
a = make_df('AB',  [1,2])
b = make_df('AB',  [3,4])
```

```python
#so now we are matching the indices 
a.index  = b.index
```

```python
pd.concat([a,b])
```

```python
#but now we have a problem that indices are repeated lets fix that thing 
```

```python
#treatng the repeated indices as the error 
```

```python
try:
    pd.concat([a,b],verify_integrity=True)
except ValueError as e:
    print(e)
```

```python
#ignorning the index 
pd.concat([a,b] ,  ignore_index=True)#this will ignore the index of the repeatinng index and will provide with the conisistent data 
```
