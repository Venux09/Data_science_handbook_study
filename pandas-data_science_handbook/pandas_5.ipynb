## Handling missing data in the pandas python 

```python
#None pythonic missing data and Null pandas missing data 
```

```python
import pandas as pd 
import numpy as np 
```

```python
val1 = np.array([1,2,None,3,4])
```

```python
val1
```

```python
for dtype in ['object','int']:
    print(f'dtype = {dtype}')
    %timeit np.arange(1E6,dtype = dtype ).sum()
    print()
```

```python
val1.sum()#such as thse aggregation function doesnotwork on differenet data types 
```

```python
val2 = np.array([1,2,3,4,np.nan])
```

```python
val2
```

```python
val2.sum()
```

```python
val2.dtype
```

```python
#Nan is like virus it affects other values also that is other than in the case 
```

```python
#Nan we can perform function while nan is present 
```

```python
#for ignoring the nan itself we can perform aggregate 
```

```python
np.nansum(val2)
```

```python
np.nanmin(val2)
```

```python
np.nanmax(val2)
```
