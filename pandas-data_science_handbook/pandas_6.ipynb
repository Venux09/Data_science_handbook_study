```python
import numpy as np 
import pandas as pd 
```

```python
#a multiply indexed series 
index = [('California', 2000), ('California', 2010),
('New York', 2000), ('New York', 2010),
('Texas', 2000), ('Texas', 2010)]
populations = [33871648, 37253956,
18976457, 19378102,
20851820, 25145561]
```

```python
pop = pd.Series(populations,index = index )
```

```python
pop
```

```python
pop[('California',2000):('Texas', 2010)]
```

```python
pop[[i for i in pop.index if i[1] == 2010]]
```

```python
pop[[i for i in pop.index if i[0] == 'California']]#this  a way of getting the desired result but it is not convient 
```

```python
#better way to do it pandas multiply indexing 
```

```python
index = pd.MultiIndex.from_tuples(index)
index
```

```python
pop = pop.reindex(index
                 )
```

```python
pop
```

```python
pop[:,2010]
```

```python
#Separating the data into two columns based on the information provided
pop_df = pop.unstack()
pop_df
```

```python
pop_df.stack()
```
