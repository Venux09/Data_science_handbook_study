```python
import pandas as pd
import yfinance as yf

# Download S&P 500 data directly into a pandas DataFrame
sp500 = yf.download('^GSPC', start='2018-01-01', end='2022-01-01')

# Preview the data
print(sp500.head())
```

```python
%matplotlib inline  
import matplotlib.pyplot as plt 
sp500.plot();
```

```python
sp500 = sp500['Close']
```

```python
sp500
```

```python
#Resampling and Converting Frequences 
```

```python
#plotting the sampling example with use of the sp500 data 
```

```python
#sampling is basically a method where we want to see the data at higher frequency from how it is originally represented
sp500.plot(alpha = 0.5,style = '-')
sp500.resample('BYE').mean().plot(style=':')
sp500.asfreq('BYE').plot(style = '--')
plt.legend(['input','resample','asfreq'] , loc = 'upper left')
```

```python
#so now we are going to to plot the freq using date since usinng date will unsample the data means it will create nan value for the 
#saturday and sunday 
```

```python
#we can solve this problem first i will observe what it will look like if we sample with data inconistencies 
fig , ax = plt.subplots(2,sharex = True)

data = sp500.iloc[:20]#taking a data sample from the entire data 

#sp500 data is for buissness days data and bussiness days data is usually from monday to friday 
data.asfreq('D').plot(ax = ax[0] , style = 'o')# i can see the inconsistency in the graph which are most probabily saturday and sunday 


data.asfreq('D',method = 'bfill').plot(ax = ax[1] ,style = 'o')
data.asfreq('D',method = 'ffill').plot(ax = ax[1], style = 'o')

ax[1].legend(['backwordfill','Forwardfill'])
```

```python
#first we resample the data in day frequency 
sp500 = sp500.asfreq('D',method='pad') #samplinng the data as Day to Day 
ROI = 100*(sp500.shift(-365) - sp500)/sp500 


ROI.plot()
plt.ylabel('Return on investment after 1 year')
```

```python
#the third type of datatime specific in pandas - Rolling windows 
import numpy as np
rolling = sp500.rolling(365,center = True)


data = pd.DataFrame({'input' : sp500,
                    'one-year rolling mean':rolling.mean(),
                    'one-year rolling median':rolling.median()},index = sp500['Ticker'])
ax = data.plot(style = ['-','--',':']  )
ax.lines[0].set_alpha(0.5)
```
