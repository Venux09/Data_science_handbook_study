```python
import numpy as np 
import pandas as pd 
class display(object):
    """Display HTML representation of multiple objects"""
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}{1}
    """
    def __init__(self, *args):
        self.args = args
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
        for a in self.args)
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
        for a in self.args)
```

### groupby : split , apply , combine 

```python
df = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
'data': range(6)},columns = ['key','data'])
```

```python
df.groupby('key').sum()#summinng the key values which are common 
```

```python
df.groupby('key').mean()
```

```python
df.groupby('key')
```

```python
#groupby funciton is special it is not viewd as Dataframe but as the DataFrameGroupBy 
```

```python
#groupby object 
```

```python
#basic group by function 
planets.groupby('method'),planets.groupby('method')['orbital_period']
```

```python
planets.groupby('method')['orbital_period'].median()
```

```python
planets.groupby('method').median()
```

```python
#group by functioin direct iteration over the groups we can apply thing on that 
#code 
for (method , group) in  planets.groupby('method') :
    
    print("{0:30s} shape = {1}".format(method,group.shape))
```

```python
#dispatch method - basically means that we can apply some of data frame objects fucnotin on the grouped object 
#means where we applied groupby 
```

```python
planets.groupby('method')['year'].describe().unstack()
```

```python
#aggregate , filter ,  transform and apply in the group by other than basic numpy operatoin and iteration , dispachig etc 
```

```python
rng = np.random.RandomState(42)
df = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'data1':range(6),
                   'data2': rng.randint(0,6,6)},
                   columns = ['key','data1','data2'])
```

```python
df
```

```python
#aggregate
```

```python
df.groupby('key').aggregate(['min',np.median,'max'])
```

```python
df.groupby('key').aggregate({'data1':'min',
                            'data2':'max'})
```

```python
#using filter in the pandas 
def filter_func (x):
    "filter func where std should be more than 4"
    return x['data2'].std()>3


display('df',' df.groupby("key").std()',' df.groupby("key").filter(filter_func)')
```
