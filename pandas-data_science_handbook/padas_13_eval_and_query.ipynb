```python
#lets do the computation using the eval 
import pandas as pd 
nrows , ncoloumns = 1000000,100
df1,df2,df3,df4 = (pd.DataFrame(rng.random((nrows,ncoloumns))) for i in range (4))
#doing simple numpy style computation 
%timeit df1+df2+df3+df4
```

```python
#doing the same computation using the eval of the pandas 

%timeit pd.eval('df1  + df2 + df3 + df4')
```

```python
#pd.eval performed the funcction 2x faster that is the big thing when you work with the billions of the rows 
```

```python
#integers 
df1, df2, df3, df4, df5 = (pd.DataFrame(rng.randint(0,1000,(100,3))) for i in range (5))
```

```python
#doing the arithmatic operators 
result_1 = -df1*df2/(df1*df2)
result_2 = df1 + df2 + df3 + df4/(df1 + df2+ df3+ df4)
```

```python
np.allclose(result_1,result_2)#since they are not equal on the terms 
```

```python
#comparison operators 
result_1 = (df1<df2) & (df2<=df3) & (df3!=df4)
result_2 = pd.eval('df1<df2 <= df3 != df4')
```

```python
np.allclose(result_1,result_2)
```

```python
#bitwise operators & and |
result_3 = (df1 < 0.5) & (df2 < 0.5) | (df3 < df4)

result_4 = ('(df1 < 0.5) & (df2 < 0.5) | (df3 < df4)')
```
