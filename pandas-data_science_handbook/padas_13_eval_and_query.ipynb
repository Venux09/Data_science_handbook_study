```python
#lets do the computation using the eval 
import pandas as pd 
nrows , ncoloumns = 1000000,100
df1,df2,df3,df4 = (pd.DataFrame(rng.random((nrows,ncoloumns))) for i in range (4))
#doing simple numpy style computation 
%timeit df1+df2+df3+df4
```

```python
#doing the same computation using the eval of the pandas 

%timeit pd.eval('df1+df2+df3+df4')
```
