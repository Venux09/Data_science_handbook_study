```python
#basic Linear Regression with scipy 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression


#reading the data 
df = pd.read_csv('data_lin.csv' , delimiter= ",")
X = df.values[:,:-1]#basically all rows:columns but only last column
Y = df.values[:,-1]#basicall all rows and last columns 

#fitting the line to the points 
fit = LinearRegression().fit(X,Y)

m = fit.coef_.flatten()
b = fit.intercept_.flatten()


#chart of this linear Regression model 
plt.plot(Y , Y , 'o')#plotting the scatter plot of the relation between the x and y 
plt.plot(X , m*X+b)#here bascially we are plotting the straight line or plotting the line which have x as plot whatever way you cann understand 
plt.show()
```

```python
#calculating the Residuals 
import pandas as pd 
import numpy as np 


points = pd.read_csv('https://bit.ly/3goOAnt' , delimiter=",").itertuples()

#slope and y intercept value of the regression model 
m = 1.93939
b = 4.73333

#calulating the residuals 
for p in points:
    y_actual  = p.y  #basically here y is values of the y column
    y_predict = m*p.x+b #and here x is the values of the x column
    residual  = y_actual-y_predict
    print(residual)
    
```
