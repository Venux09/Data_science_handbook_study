```python
#finding the values of m and b using the closed form equation where we have one input and one variable 
import pandas as pd 


#load the data 
points = list(pd.read_csv('https://bit.ly/2KF29Bd' , delimiter  = ",").itertuples())


#writing the formula for the closed equation 

n = len(points)

#this is the formula for finding the slope or m whatever
m = (n*sum(p.x*p.y for p in points) - (sum(p.x for p in points)*sum(p.y for p in points)))/(n*sum(p.x**2 for p in points ) - sum(p.x**2 for p in points))

#formula for the finding the b values 
b = (sum(p.y for p in points)/n) - m*sum(p.x for p in points)/n
```

```python
import pandas as pd 
import numpy as np
from numpy.linalg import inv ,qr
```

```python
#importing the data 
df = pd.read_csv('https://bit.ly/3goOAnt', delimiter=",")

#extracting all the column all the rows but last columns 
X = df.values[:,:-1].flatten()

#making a place holder like 1 that we make in order to solve an inverse 
X_1 = np.vstack([X,np.ones(len(X))]).transpose()

#extracting the output values from the data means that results whatever 
Y = df.values[: , -1]


# calculate coefficents for slope and intercept
# using QR decomposition
Q, R = qr(X_1)
b = inv(R).dot(Q.transpose()).dot(Y)

print(b)
```

```python
import random 


def f(x):
    return (x-3)**2+4

def dx_f(x):
    return 2*(x-3)

#The Learning Rate 
L = 0.001

#the number of iteration 
iterations = 100_000


#starting the random x 
x = random.randint(-15 , 15)


for i in range(iterations):
    d_x = dx_f(x)

    x -= d_x*L

print(x,f(x))
```

```python
#performing Gradient Decent for the Linear Regression
```

```python
import pandas as pd 

#data 
points = list(pd.read_csv("https://bit.ly/2KF29Bd").itertuples())

#Building a model 
m = 0
b = 0 

#learning Rate 
L = 0.001 

#iteration mean how much the program will run 
iterations = 100_000 

#Number of points in the data x taking its lenght 
n  = float(len(points))


#performing the gradient decent 
for i in range (iterations):
    #slope toward m 
    D_m = sum(2*p.x*((m*p.x+b)-p.y) for p in points)

    #slope toward b 
    D_b = sum(2*((m*p.x+b) - p.y) for p in points)

    #updating the m and b 
    m -= L*D_m
    b -= L*D_b

print (f'Y = {m} + {b}')
```

```python
#calculating the partial derivatives using sympy 
from sympy import *

m , b , i , n = symbols('m b i n')
x , y = symbols ('x y' , cls=Function)

sum_of_squares = Sum((m*x(i) + b - y(i)**2) , (i , 0 , n))#here we are finding the square of the subtracting the x and errors values 


#applying the differentiation on the both m and b 
d_m = diff(sum_of_squares,m)
d_b = diff(sum_of_squares,b)
```
