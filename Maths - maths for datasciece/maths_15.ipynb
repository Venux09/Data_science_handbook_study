```python
#converting z score to x values and vice versa mean converting x values to z scores 
```

```python
def z_score(mean,std_dev,x):
    "This function is basically based on z score formula"
    return (x - mean)/std_dev

def z_score_to_x(mean,std_dev,z_score):
    "This function changes the z score to x which is a original value"
    return ((z_score*std_dev)+mean)
```

```python
#solving a z score as well as doing the vice versa 
mean = 140000
std_dev = 3000
x = 150000
z_score = z_score(mean,std_dev,x)
z_score  #how spread out of the data is 
```

```python
z_to_x = z_score_to_x(mean,std_dev,z_score)
z_to_x 
```
