### Binomial distribution from the scracth 

```python
#building a binomial distribution 
def factorial(n:int):#so here we are making a factorial function 
    f = 1
    for i in range(n):
        f *= i+1
    return f 


#function for finding the pnc of the given number 
def binomial_coefficient(n:int,k:int):
    return factorial(n)/(factorial(k)*factorial(n-k))


#finding the binomial distribution usinng trials

def binomial_distribution(n:int,k:int,p:float):
    #we can use varibale:[] for making a value in certain binary 
    return binomial_coefficient(n,k)*(p**k)*(1-p)**(n-k)#it is the binomial distribution formula just mean that distribution * probabiliy of all success and  probabilty of the all failure 
    
    
```

```python
%xmode Verbose
#finding the binomial distribution using the formula's 
n = 10 
p = 0.9 
for k in range(n+1):
    binom = binomial_distribution(k,n,p)
    print("{0}-{1}".format(k,binom))
```
