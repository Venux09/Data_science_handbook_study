```python
#a more complicated vector transformation 


#declaring the i hat and j hat 
i_hat = array([2,1])
j_hat = array([2,-3])


#composing the vector and as well making the rows and the columns 
basis = array([i_hat , j_hat]).transpose()
```

```python
#declaring the vector v  
v = array([2,1])


#making the 1 new vector using the product of the basis on the basis of the v 
new_vector = basis.dot(v)
```

```python
new_vector
```

```python
#combining two transformation 
import numpy as np 

#Tranformation - 1

#declaring the i_hat and j_hat 
i_hat1 = array([1,2])
j_hat1 = array([2,3])

#transforming the Tranformation - 1 
basis_1 = array([i_hat1,j_hat1]).transpose()

#Tranformation - 2 

#declaring the i_hat and j_hat 
i_hat2 = array([1,3])
j_hat2 = array([2,3])

#transforming the Transformation -2 
basis_2 = array([i_hat2 , j_hat2]).transpose()



#combining the two transpose or Transformation
combined = basis_1@basis_2  #@ - operator used for making the transform 

#declaring the vector 
v = array([1,1])

#making the new vector from the combined Transformation
new_vector = combined.dot(v)
```

```python
new_vector
```
