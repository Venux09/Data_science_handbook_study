# Multiple Subplots in the data 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
%matplotlib inline 
plt.style.use('dark_background')
```

```python
ax1 = plt.axes()#standart size means a normal plane graph 
ax2 = plt.axes([0.65, 0.65, 0.2, 0.2])#we are just customizing the size of the planar in order so that we could change the size object
ax3 = plt.axes([0.80, 0.80, 0.4, 0.4])
```

```python
#here we are just plotting the same stuff but here we are doing a thing scalled putting values into that function 

fig = plt.figure()
ax1 = fig.add_axes([0.1, 0.5, 0.8, 0.4],xticklabels=[], ylim=(-1.2, 1.2))
ax2 = fig.add_axes([0.1, 0.1, 0.8, 0.4],ylim=(-1.2 , 1.2))

x = np.linspace(0 , 10)
ax1.plot(np.sin(x))
ax2.plot(np.cos(x))
```

```python
for i in range(1,7):
    plt.subplot(2,3,i)#plotting 8 plots at once using subplots 
    plt.text(0.5 , 0.5 , str((2,3,i)) , ha = 'center')
```

```python
fig, ax = plt.subplots(2, 3, sharex='col', sharey='row')
```

```python
fig , ax = plt.subplots(2 , 3 , sharex = 'col' , sharey = 'row')
```

```python
grid = plt.GridSpec(2 , 3 , wspace = 0.4 , hspace = 0.3)
plt.subplot(grid[0, 0])
plt.subplot(grid[0, 1:])
plt.subplot(grid[1, :2])
plt.subplot(grid[1 , 2])
```
