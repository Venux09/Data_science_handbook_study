```python
%matplotlib inline 
import matplotlib.pyplot as plt 
import numpy as np 

plt.style.use('dark_background')
```

```python
#Scatter plot with using the plt.plot 
x = np.linspace(0,10,30)
y = np.sin(x)

plt.plot(x,y,'o', color = 'blue')
```

```python
#lets see how different type fo marker there are 

rng = np.random.default_rng(0)#settinng the threshold 

for marker in ['o', '.', ',', 'x', '+', 'v', '^', '<', '>', 's', 'd']:
    plt.plot(rng.random(2) , rng.random(2) , marker , color = 'yellow',label = 'marker'.format(marker))

plt.legend(numpoints = 1 ,fontsize = 13)
plt.xlim(0 , 1.8)
```

```python
#making more specification with the marker which are fun to do around 
plt.plot(x,y,'-p' , color  = 'red'  , markersize = 14 , markerfacecolor  = 'grey' , markeredgecolor = 'blue', linewidth  = 4 , 
         markeredgewidth = 2)#this way we can change how 
```

```python
#using plt.scatter to make the plots 
```

```python
plt.scatter(x,y,marker = 'o')
```

```python
#Random plots with many colors and randomm sizes
rng = np.random.default_rng(0)
x = rng.normal(size =100)

y = rng.normal(size = 100)

colors = rng.random(100)

sizes = 1000*rng.random(100)

plt.scatter(x,y,c= colors, s= sizes , alpha = 0.3)
plt.colorbar()
```

```python
#importing the iris data sets in order to learn about the color bar and scatter more 
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
features = iris.data.T
```

```python
plt.scatter(features[0] , features[1] , s = 100*features[3] , alpha = 0.5 , c = iris.target , cmap = 'viridis' )
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
```

```python
#Visualizing the Uncertainities 
%matplotlib inline 

import matplotlib.pyplot as plt 
import numpy as np 

x = np.linspace(0,10,50)
dy = 0.8
y = np.sin(x)+dy*np.random.randn(50)


#plotting the errorbar 
plt.errorbar(x,y,yerr=dy , fmt='.k' , color = 'yellow' , ecolor= 'grey' , elinewidth = 3 , capsize = 0)#so fmt here is just for plotting style 
plt.grid(True)
```

```python
#making the graph of the continuity of the errors 
from sklearn.gaussian_process import GaussianProcessRegressor

from sklearn.gaussian_process import GaussianProcessRegressor
# define the model and draw some data
model = lambda x: x * np.sin(x)
xdata = np.array([1, 3, 5, 6, 8])
ydata = model(xdata)
# Compute the Gaussian process fit
gp = GaussianProcessRegressor()
gp.fit(xdata[:, np.newaxis], ydata)
xfit = np.linspace(0, 10, 1000)
yfit, dyfit = gp.predict(xfit[:, np.newaxis], return_std=True)
```

```python
#visulizing the result 
plt.plot(xdata , ydata, 'or')
plt.plot(xfit,yfit , '-' , color = 'gray')
plt.fill_between(xfit , yfit-xfit , yfit+xfit , color = 'gray' , alpha = 0.2)
plt.xlim(0 , 10)
```
