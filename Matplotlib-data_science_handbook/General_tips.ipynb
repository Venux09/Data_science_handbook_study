# General tips for matplotlib

```python
import matplotlib as mlp 
import matplotlib.pyplot as plt
import numpy as np
```

```python
%matplotlib inline 

x = np.linspace(0,10,100)
fig = plt.figure()#it is  a command to save the picture you are making 
plt.plot(x,np.sin(x) , '-')
plt.plot(x,np.cos(x) , '--')
```

```python
fig.savefig('my_fig.png')
```

```python
from IPython.display import Image 
Image('my_fig.png')#to call the image in your directory 
```

```python
#Creating the two interface matlab style interface 
plt.subplot(2,1,1)#rows,columns,panel_number 
plt.plot(x,np.cos(x) , '-')

plt.subplot(2,1,2)
plt.plot(x,np.sin(x)  , '--')
```

```python
#object oriented - interface 
fig,ax = plt.subplots(2)

#the previous one was not object oriented in that there is difficullity in making changes when it matters 
ax[0].plot(np.sin(x),'-')
ax[1].plot(np.cos(x),'+')
```
