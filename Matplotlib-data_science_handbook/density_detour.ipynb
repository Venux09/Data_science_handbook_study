# Density and Contour Plots

```python
%matplotlib inline 
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd  
plt.style.use('dark_background')
```

```python
#lets start plotting some color coded contour 
```

```python
def f(x,y):
    "Defining the values of the function that we will make out of our raw x and y values "
    return np.sin(x)**10+np.cos(10+x*y)*np.cos(x) 


#value of x and y 
x = np.linspace(0,5,50)
y = np.linspace(0,5,50)
X,Y = np.meshgrid(x,y)#meshgrid what i searched means that it change our 1dimensional array to 2D relation that can happen at the graph 
Z = f(X,Y) #now we will make the Z function which is basically functon we have putted on the vales of the x and y 
```

```python
#plotting the Contour contour fuction 
plt.contour(X,Y,Z , cmap = 'RdGy')
```
