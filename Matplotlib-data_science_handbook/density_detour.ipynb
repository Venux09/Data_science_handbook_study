# Density and Contour Plots

```python
%matplotlib inline 
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd  
plt.style.use('seaborn-v0_8-white')
```

```python
#lets start plotting some color coded contour 
plt.style.available
```

```python
def f(x,y):
    "Defining the values of the function that we will make out of our raw x and y values "
    return np.sin(x)**10+np.cos(10+x*y)*np.cos(x) 


#value of x and y 
x = np.linspace(0,5,50)
y = np.linspace(0,5,50)
X,Y = np.meshgrid(x,y)#meshgrid what i searched means that it change our 1dimensional array to 2D relation that can happen at the graph 
Z = f(X,Y) #now we will make the Z function which is basically functon we have putted on the vales of the x and y 
```

```python
#plotting the Contour contour fuction 
plt.contour(X,Y,Z, cmap = 'RdGy' )
```

```python
#option that we have in the color map 
# plt.cm.
```

```python
#plotting it with the color bar as well 
plt.contour(X,Y,Z , cmap = 'RdGy')
plt.colorbar()
```

```python
#this graph look like it have so much descrete values lets make is more uniform so we can get a proper scence to choose sume result on the set of the values 
```

```python
#plt.imshow does not accept the x and y so we have given extent of x and y 
#

plt.imshow(Z , extent = [0,5,0,5] , origin = 'lower' , cmap = 'RdGy' , interpolation='Gaussian',aspect = 'equal')
plt.colorbar()
```
