# Learning to design the plot legends 

```python
import numpy  as np 
import pandas as pd 
import matplotlib.pyplot as plt 
plt.style.use('dark_background')
```

```python
#lets plot the legends
x = np.linspace(0,10,1000)

fig , ax = plt.subplots()
ax.plot(x , np.sin(x) , '-' , label = 'sine')
ax.plot(x , np.cos(x) , '*',color='yellow' , label  = 'cose')

ax.axis('equal')
ax.legend(loc = 'upper left' , frameon = True , shadow = True , borderpad = 1 , fancybox = True )
```

```python
y = np.sin(x[:, np.newaxis] + np.pi * np.arange(0, 2, 0.5))
lines = plt.plot(x , y )

plt.legend(lines[:2] , ['first' , 'second'] , frameon = True)
```

```python
#this was not intutive lets plots on the old style 
plt.plot(x , y[:,0] , label = 'first'  )
plt.plot(x , y[:,1] , label = 'second'  )
plt.plot(x , y[: ,2])
plt.legend(frameon = True)
```

```python
#multiple  legends on the graph 
fig , ax = plt.subplots()

lines = []
styles = ['-' ,'--' , '-.' , ':']
x = np.linspace(0 , 10 , 1000)


for i in range(4):
    lines += ax.plot(x, np.sin(x - i * np.pi / 2),styles[i] , color = 'black')

ax.axis('equal')

#specifying the legends for the first one 
ax.legend(lines[:2] , ['first' , 'second'] , loc = 'upper right')


#now making the legend for the second ones 
from matplotlib.legend import Legend 
leg = Legend(lines[2:] ,   ['third' , 'fourth'] , loc = 'lower right')
ax.add_artist(leg);
 
```
