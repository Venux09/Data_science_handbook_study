# Colorbars

```python
import matplotlib.pyplot as plt 
plt.style.use('dark_background')
```

```python
import numpy as np 
x = np.linspace(0, 10, 1000)
I = np.sin(x) * np.cos(x[:, np.newaxis])
```

```python
plt.imshow(I,cmap='Blues')
plt.colorbar()
```

```python
plt.imshow(I,cmap ='berlin' )
```

```python
#we need color extents and limits because sometime we can get data types where we want to perform noise so that every number can get different 
#color of the data that can help to find the desired result out of the data 
```

```python
#make a noise in 1% image pixels 
speckles = (np.random.random(I.shape) < 0.01) 
I[speckles] = np.random.normal(0 , 3 , np.count_nonzero(speckles))

#plotting the graph 
plt.figure(figsize = (10 , 3.5))

plt.subplot(1,2,1)
plt.imshow(I , cmap = 'RdBu')
plt.colorbar()

plt.subplot(1,2,2)
plt.imshow(I , cmap = 'RdBu')
plt.colorbar(extend = 'both')
plt.clim(-1,1)
```
