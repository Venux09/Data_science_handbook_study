```python
%matplotlib inline 
import matplotlib.pyplot as plt 
import numpy as np 

plt.style.use('dark_background')
fig = plt.figure() #here we can start to make the figure 
ax = plt.axes() #axes is a term which is refered the box which is under right now

x = np.linspace(0,10,1000)#assining the value of the x 
ax.plot(x,np.sin(x))
plt.plot(x,np.sin(x-1) , color = 'blue') #setting the color on the behalf of the name
plt.plot(x,np.sin(x-2) , color = 'b')#on the basis of the character 
plt.plot(x,np.sin(x-3), color = '0.75')#on the basis of the gray meter 
plt.plot(x, np.sin(x-4) , color = '#FFDD44')#on the basis of hex code setting up a different color 

plt.plot(x,np.sin(x-6), color = 'chartreuse')#html color names supported 
```

```python
%matplotlib inline 
import matplotlib.pyplot as plt 
import numpy as np 
#different linestyle in the matplot 
plt.style.use('dark_background')
fig = plt.figure() #here we can start to make the figure 
ax = plt.axes() #axes is a term which is refered the box which is under right now

x = np.linspace(0,10,1000)#assining the value of the x 
ax.plot(x,np.sin(x))
plt.plot(x,np.sin(x-1) , color = 'blue') #setting the color on the behalf of the name
plt.plot(x,np.sin(x-2) , color = 'b')#on the basis of the character 
plt.plot(x,np.sin(x-3), color = '0.75')#on the basis of the gray meter 
plt.plot(x, np.sin(x-4) , color = '#FFDD44')#on the basis of hex code setting up a different color 

plt.plot(x,np.sin(x-6), color = 'chartreuse')#html color names supported 
```

```python
#differen linestyle in the matplot lib 
plt.plot(x,x+0,linestyle = '-')#solid style 
plt.plot(x,x+1,linestyle = '--')#dashed style 
plt.plot(x,x+2 , linestyle = '-.')#dash shot - it is something new 
plt.plot(x,x+3,linestyle = ':')#dotted style 
```

```python
#labeling plots 
plt.plot(x,np.sin(x))
plt.title('Simple sin curve')
plt.xlabel('x')
plt.ylabel('sin(x)')
```

```python
#using multiple plots in one axes 
plt.plot(x,np.sin(x) , ':b' , label = 'sin(x)')
plt.plot(x , np.cos(x) , '-g' , label = 'cos(x)')
plt.axis('equal')
#using plt.legend to reresent thhe name of the plots 
plt.legend()
```

```python
#using the object oriented we can apply all these things at once rather than calling plt all the time 
ax = plt.axes()
ax.plot(x,np.sin(x))
ax.set(xlim = (0,10) , ylabel = 'sin(x)' , xlabel = 'values') 
```
