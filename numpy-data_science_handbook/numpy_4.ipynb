### Array Concatenation and Splitting
All of the preceding routines worked on single arrays. It’s also possible to combine
multiple arrays into one, and to conversely split a single array into multiple arrays.

```python
x1 = np.random.randint(10,size =(2,3))
x2 = np.random.randint(10,size =(2,3))
x3 = np.random.randint(19000,size =(2,3))
```

```python
#concatenating multiple array one can be at any limit 
print(np.concatenate([x1,x2,x3,x2,x3,x2,x1,x1]))#it is concatenated in the firsts exis which is a defalut exis by far 
```

```python
print(np.concatenate([x1,x2,x3,x2,x3,x2,x1,x1],axis=1))#concatenating in the axis = 1 
```

```python
#axis 0 means it is going downward - vertically 
#axis 1 means it is going sidebyside - horizontal 
```

```python
#vstacking in numpy means vertical stacking 
a = np.arange(1,4)
a1 = np.random.randint(4,size=(2,3))
```

```python
%xmode Verbose
print(np.vstack([a,a1]))
```
