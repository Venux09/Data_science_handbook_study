# Structured data: Numpy's structured arrays 


```python
name = ['Alice', 'Bob', 'Cathy', 'Doug']
age = [25, 45, 37, 19]
weight = [55.0, 85.5, 68.0, 61.5]
```

```python
#we have created our own array to fit in the in the things as the data types 
import numpy as np 
data = np.zeros(4,dtype={'names':('name','age','weight'),
                        'formats':('U10','i4','f8')})
```

```python
data.dtype
```

```python
data['name'] = name
data['age']  = age 
data['weight'] = weight
```

```python
data['name']
```

```python
data
```

```python
data[0]
```

```python
data[['name','age']]
```

```python
data[data['age']<30]['name']
```
