```python
# We can think of this as an operation that stretches or duplicates the value 5 into the
# array [5, 5, 5], and adds the results. The advantage of NumPy’s broadcasting is that
# this duplication of values does not actually take place, but it is a useful mental model
# as we think about broadcasting.
```

```python
p = np.ones((3,2))
q = np.arange(3)
```

```python
#so here it will give error becuase there is one column missing inthe the p because to broadcaste over here there will 
#hence , it can't be broadcasted here
```

```python
p+q
```

```python
#solving the problem
r = p.copy()
```

```python
#right side padding 
r = r.reshape(2,3)
```

```python
r+q
```

```python
#so the problem is solved by changing the shape of the array accorind the shape required to fit in this |
```

```python
#this broadcasting is also called padding the values in the left side to match the shape of the arrays
```

```python
#left side padding of the array
#using the newaxis formula to do the left side padding of the array 
p+q[:,np.newaxis]
```

```python
#--- BROADCASTING CAN BE APPLIED TO ANY BINARY FUNCTION INCLUDING LOG EXP
```
