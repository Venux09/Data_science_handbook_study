## Fancy indexing in numpy 

```python
#let's generate a random array 
rand = np.random.RandomState(42)#just 42 RANDOM STATE  blaah just 42 is joke 
a = rand.randint(100, size = 10)
```

```python
a
```

```python
#normal indexing  
a[0]
```

```python
a[2]
```

```python
inc = [2,3,4]
a[inc]
```

```python
#or something like this 
[a[0],a[1]]
```

```python
#here comes fancy indexing 
```

```python
b = np.array([[2,3],[4,5]])
a[b]
```

```python
ind = np.array([[2,3,4],
    [4,5,6],
    [1,6,2]])
```

```python
a[ind]
```

```python
#this is cool because here we are not generating array for the ind only it is an actualy array and we are doing indexing on that actual array usingn other data 
```

```python
#combined indexing 
#in combined indexinng we are going to use both indexking and fancy indexkinng 
```

```python
#lets generate an array 

a = np.arange(9).reshape(3,3)
a 
```

```python
a[2,[1,2,0]]+a[0,[0,1,2]]# type of mixing the broadcasting  , normal indexing, fancy indexing 
```

```python
a[:,[1,2,0]]#indexes with slices 
```

```python
a[1:2 ,[0,1,2]]#second row of the column 
```

### Example of using the fancy indexing , common indexing and slicing 

```python
#One common use of fancy indexing is the selection of subsets of rows from a matrix.
#For example, we might have an N by D matrix representing N points in D dimensions,
#such as the following points drawn from a two-dimensional normal distribution:
```

```python
mean = [0,0]#mean of the attributes given 
cov = [[1,2],[2,5]]#covariance matrix means how much they are related to each other given that these are the relashions 
x = rand.multivariate_normal(mean,cov,100)
x.shape
```

```python
#so here we are plotting the normally distributed graph using it points relashion with columns and  rows  
```

```python
%matplotlib inline 
import matplotlib.pyplot as plt
import seaborn ; seaborn.set()

plt.scatter(x[:,0],x[:,1])#so here we are just plotting the relation given that we are taking all the value of the columsn and rows 
```
