## Fancy indexing in numpy 

```python
#let's generate a random array 
rand = np.random.RandomState(42)#just 42 RANDOM STATE  blaah just 42 is joke 
a = rand.randint(100, size = 10)
```

```python
a
```

```python
#normal indexing  
a[0]
```

```python
a[2]
```

```python
inc = [2,3,4]
a[inc]
```

```python
#or something like this 
[a[0],a[1]]
```

```python
#here comes fancy indexing 
```

```python
b = np.array([[2,3],[4,5]])
a[b]
```

```python
ind = np.array([[2,3,4],
    [4,5,6],
    [1,6,2]])
```

```python
a[ind]
```

```python
#this is cool because here we are not generating array for the ind only it is an actualy array and we are doing indexing on that actual array usingn other data 
```

```python
#combined indexing 
#in combined indexinng we are going to use both indexking and fancy indexkinng 
```

```python
#lets generate an array 

a = np.arange(9).reshape(3,3)
a 
```

```python
a[2,[1,2,0]]+a[0,[0,1,2]]# type of mixing the broadcasting  , normal indexing, fancy indexing 
```

```python
a[:,[1,2,0]]#indexes with slices 
```

```python
a[1:3 ,[0,1,2]]
```
