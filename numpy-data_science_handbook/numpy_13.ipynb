## Using this to binning the data using the modification using fancy indexing 


```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn ; seaborn.set()

np.random.seed(42)
#creating an array 
x = np.random.randn(100)

bins = np.linspace(-5,5,20)
counts = np.zeros_like(bins)


#histogram by hands 
i = np.searchsorted(bins,x)

#adding 1 to each line 
np.add.at(counts,i,1)

#plotting histogram 
plt.plot(bins,counts,linestyle='-')
```

## Sorting in numpy and several algorithms 

```python
import numpy as np 
def selection_sort(x):
    for i in range (len(x)):
        swap = i+np.argmin(x[i:])
        (x[i],x[swap]) = (x[swap],x[i])
    return x ,print(f'swap - {swap}')#swapping the value from postion to position

x = np.array([1,53,43,9,3])
selection_sort(x)
```

```python
def bogosort(x):
    while np.any(x[:-1]>x[1:]):
        np.random.shuffle(x)
    return x 

x = np.array([1,4,3,5,7,0,9,8])
bogosort(x)
```
