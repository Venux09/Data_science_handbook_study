```python
import numpy as np 
x = np.arange(10)
a = [2,3,4,5]
x[a] = 0 #it is a way of modying multiple value at once 
x
```

```python
y = np.zeros(10)
y[[0,0]] = [4,6]
y#here we can see that 4 is not in the result it is simply means that first it becomes 4 then i comes to 6 and becomes 6 thats why it is not  4
```

```python
z = np.zeros(10)
i = [2,3,3,4,4,4]
z[i] += 1 #here we were imagening that it will give 1 , 2, 3 but we were wrong it is just performing onnn bases of i 
z
```

```python
#to perform it correctly 
np.add.at(z,i,1)
z
```
