### Computation on NumPy Arrays: Universal Functions

```python
#lets caluculate the reciprocal of the numbers
```

```python

import numpy as np 
np.random.seed(0)

def computing_reciprocal(values):
    ouput = np.empty(len(values))
    for i in range (len(values)):
        ouput[i] = 1/values[i]

    return ouput

values = np.random.randint(1,11,size = 5)
computing_reciprocal(values)
```

```python
big_array = np.random.randint(1,100,size=10000)

%timeit computing_reciprocal(big_array)
```
