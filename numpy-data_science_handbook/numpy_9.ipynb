```python
import pandas as pd 
data = pd.read_csv('presidents_height.csv')
```

```python
heights = np.array(data['height(cm)'])
```

```python
heights
```

### data about the presidents height of the usa 

```python
f'Averga height :- {np.mean(heights)}'
f'Standart deviation of the heights :- {np.std(heights)}'
f'maximum heights:- {np.max(heights)}'
f'minimum heights :- {np.min(heights)}'
```

```python
f'percentile(99):- {np.percentile(heights,99)}'
```

```python
f'percentile(50)-{np.percentile(heights,50)}'
```

```python
f'median :- {np.median(heights)}'
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn; seaborn.set()
plt.hist(heights)
plt.title('Us president Heights data ')
plt.xlabel('Height(cm)')
plt.ylabel('Numbers')
```
