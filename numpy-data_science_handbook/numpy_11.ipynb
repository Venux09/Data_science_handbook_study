```python
#centering the array means making it's mean zero and it shows how far is one value from its center 

#example 
import numpy as np 
x = np.random.random((10,3))
xmean = x.mean(0)
x_centered = x-xmean #here is the centered array of the version of x 
```

```python
x_centered#it shows how far the value is from it's actual mean values 
```

```python
#plotting of the 2 - dimensional functions 
x = np.linspace(0,5,50)
y = np.linspace(0,5,50)[:,np.newaxis]

#function z = f(x,y)
z = np.sin(x)**10+np.cos(10+x*y)*np.cos(x)
z.shape
```

```python
#to plot this 2 dimensional broadcasted array  array i will use matplotlib
%matplotlib inline
import matplotlib.pyplot as plt 
plt.imshow(z,origin='lower',extent = [0,5,0,5],cmap='viridis')
plt.colorbar()
```
