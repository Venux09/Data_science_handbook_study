### %prun Profiling the scrip in jupyter notebooks 

```python
def sum_of_lists(N):
    total = 0
    for i in range(5):
        L = [j^(j>>i) for j in range(N)]
        total += sum(L)
    return total
```

```python
%prun sum_of_lists(100000)
```
