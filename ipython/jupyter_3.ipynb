### debugging tools in jupyter

```python
def func1(a,b):
    "Return the division of of a by b"
    return a/b

def func2(x):
    a = x 
    b = x-1 
    return  func1(a,b)

```

```python
#%xmode have three mode of errors ouput Plain , verbose , Context 
```

```python
%xmode Plain
```

```python
func2(1)
```

```python
%xmode Verbose
```

```python
func2(1)#verbose give more context about the error than any mode of %xmode
```
