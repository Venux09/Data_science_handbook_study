```python
from scipy.stats import norm 
def critical_z_value(p):
    norm_dist = norm(loc = 0 , scale = 1)#scalling the values from 0 to 1 
    tail_area = 1-p/2 #dividing by 2 because it will make area for tail and upper equal 
    upper_area = 1  - (1-p/2) # this is upper tail area 
    return norm_dist.ppf(tail_area),norm_dist.ppf(upper_area)
```

```python
critical_z_value(p = 0.95)
```
