```python
#median is the most centermost value of the data if the data is in the even number then there will be two middle most values
```

```python
#calculating the median no of pets the owner owns 
samples = [1,2,5,6,7,9,8]

def median(values):
    ordered = sorted(values)#sorting the values in the numberwise 
    n = len(ordered)
    mid = int(n/2) - 1 if n%2 == 0 else int(n/2)#it is giving the value of the index of the median since index inn the python start from 0

    if n%2 == 0 :
        return (ordered[mid]+ordered[mid+1])/2#it give the two middle values 

    else :
        return ordered[mid]


```

```python
median(samples)
```

```python
# #mode 
# The mode is the most frequently occurring set of values. It primarily becomes
# useful when your data is repetitive and you want to find which values occur the
# most frequently.
```

```python
#when no value occur in the data more than once there is no mode and when 2 values  and have equal frequency of the repetetion than 
#that is called bimodal 
```

```python
#mode formula in the python 
from collections import defaultdict 

sample = [1, 3, 2, 5, 7, 0, 2, 3,3,2]#number of pets each owner owns 


def mode(values):
    counts = defaultdict(lambda:0)#creates a empty dictionary 


    for s in values:
        counts[s] += 1 

    max_counts  = max(counts.values())
    modes = [v for v in set(values) if counts[v] == max_counts]
    return modes 


mode(sample)
```

```python
#population differences and standart deviation 
#step deviatiation basically means that how spread out our data is , it tell the how far the values is from their means and is a very 
#important tool in the statistics 
```

```python
#calculating the  variance of the data of the our pets which owners owns 

data = [0, 1, 5, 7, 9, 10, 14]#no of the pets each owner owns 


def variance(values):
    mean = sum(values)/len(values)#mean formula in the python 

    _variance = (sum((v-mean)**2 for v in values)/len(values))#variance formula in the python 

    return _variance 
from math import sqrt 
def step_deviation(values):
    return sqrt(variance(values))
```

```python
variance(data)
```

```python
step_deviation(data)
```

```python
#calculating the  variance of the data of the our pets which owners owns 

data = [0, 1, 5, 7, 9, 10, 14]#no of the pets each owner owns 


def variance(values,is_sample : bool=False):
    mean = sum(values)/len(values)#mean formula in the python 

    _variance = _variance = sum((v - mean) ** 2 for v in values) /(len(values) - (1 if is_sample else 0))#variance formula in the python 

    return _variance 
from math import sqrt 
def step_deviation(values,is_sample=False):
    return sqrt(variance(values,is_sample))#since step deviation is a sqrt of the variance 
```

```python
variance(data,is_sample = True)
```

```python
step_deviation(data,is_sample=True)
```
