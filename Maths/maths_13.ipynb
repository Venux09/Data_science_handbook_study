```python
#median is the most centermost value of the data if the data is in the even number then there will be two middle most values
```

```python
#calculating the median no of pets the owner owns 
samples = [1,2,5,6,7,9,8]

def median(values):
    ordered = sorted(values)#sorting the values in the numberwise 
    n = len(ordered)
    mid = int(n/2) - 1 if n%2 == 0 else int(n/2)#it is giving the value of the index of the median since index inn the python start from 0

    if n%2 == 0 :
        return (ordered[mid]+ordered[mid+1])/2#it give the two middle values 

    else :
        return ordered[mid]


```

```python
median(samples)
```
