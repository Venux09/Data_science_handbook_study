```python
#summation of the elements in the python 
x = [1,2,3,4,5]
n = len(x)


summation = sum(10*x[i] for i  in range (0,n))
summation 
```

```python
from sympy import *
#summation 
i , n = symbols('i n')
#iterating values from i which is 1 to n 
summation = Sum(2*i,(i,1,n))#have to use Sum not sum 

#setting n as 5 so it will iterate the value then i will multiply from 1 to n then will add it 
subs_5 = summation.subs(n,5) #setting n as 5 
subs_5.doit() # do it function for running the expression 
```
