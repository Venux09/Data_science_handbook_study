```python
from sympy import *
x , s = symbols('x s') #setting up the symbols 

#function 
f = x**2


slope_x = (f.subs(x,x+s) - f)/((x+s)-x)#function for calculating the slope of any number 
slope_2 = slope_x.subs(x,2)#substituting the value of the x as 2 
result  = limit(slope_2,s,0)#setting the limit of the of s to 0 
result
```
