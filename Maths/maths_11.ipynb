```python
#using scipy for solving the binom dist 
from scipy.stats import binom 


#number of trials
n = 10
#probability 
p = 0.9

for k in range(n+1):
    #k is the range of the number we have setted previously
    probability = binom.pmf(k,n,p)#here pmf stands for the probability mass function 
    print('{0}-{1}'.format(k,probability))
    
```
