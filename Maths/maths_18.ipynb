```python
from scipy.stats import norm 
#finding the z critical values 
def critical_z_value(p):
    norm_dist = norm(loc=0.0,scale=1.0)
    left_tail_area = (1-p)/2
    upper_area = 1 - ((1-p)/2)
    return norm_dist.ppf(left_tail_area) , norm_dist.ppf(upper_area)
```

```python
critical_z_value(p=0.95)
```

```python
from math import sqrt
#caluculating the confidence intervals in the python 
def confidence_interval(p,sample_mean,sample_deviation , n):
    lower_zc,upper_zc = critical_z_value(p)
    lower_me = lower_zc*(sample_deviation/sqrt(n))
    upper_me =upper_zc*(sample_deviation/sqrt(n))
    return sample_mean+lower_me , sample_mean+upper_me
```

```python
confidence_interval(p=.95,sample_mean =64.408,sample_deviation = 2.05 , n = 31)
```
