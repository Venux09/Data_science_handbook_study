```python
from math import sqrt
sample = [1.78, 1.75, 1.72, 1.74, 1.77]

mean = sum(sample)/len(sample)

#standart deviation:
def variance(values):
    "This functoin returns the standart deviation of the values"
    return sum((v-mean)**2 for v in values)/len(values)

def std(values):
    return sqrt(variance(values))
    
```

```python
#solving question number 3
from math import sqrt 
from scipy.stats import norm

def crtical_z(p,mean=0,std = 1):
    norm_dist = norm(loc = mean, scale = std)
    lower_area = (1-p)/2
    upper_area = 1-((1-p)/2)
    return norm_dist.ppf(lower_area),norm_dist.ppf(upper_area)

def margins(p,sample_mean,sample_std,n):#n=Value of the sampel 
    lower,upper = crtical_z(p)#p is value of which we want to find the confidence interval 
    lower_ci = lower*(sample_std/sqrt(n))
    upper_ci = upper*(sample_std/sqrt(n))
    return lower_ci+sample_mean,upper_ci+sample_mean

margins(p=0.99,sample_mean=1.715588,sample_std=0.029252,n=34)
```
